# Function Basis

```python
import os
import numpy as np
import matplotlib.pyplot as plt

FIGURES_DIR = "figures"
DATA_DIR = "data"

if not os.path.exists(FIGURES_DIR):
    os.makedirs(FIGURES_DIR)
```

```python
def save_plot(fig, filename):
    fig_file_path = os.path.join(FIGURES_DIR, filename)
    fig.savefig(fig_file_path, dpi=300)
```

```python
samples =  np.load(os.path.join(DATA_DIR, "functions.npy"))
```

# Task 1

```python
X = np.arange(101) * 0.06
f_mean = np.mean(samples, axis=0)
print(f_mean.shape)
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.set_title(r"$f^{(k)}$ mean")
ax.set_xlabel(r"$\mathcal{X}$")
ax.set_ylabel(r"$f^{(k)}(\mathcal{X})$ mean")
ax.grid(alpha=0.3)
ax.plot(X, f_mean)
plt.show()
save_plot(fig, "f_mean_plot.png")
```
